# 02 — Order-flow & CVD

Analiza buy_ratio, volume_delta, CVD, imbalance, divergencias precio vs CVD.

```python
import polars as pl, matplotlib.pyplot as plt
df = pl.read_parquet("../data/processed/btcusdt_1m.parquet")
from src.features import build_features
df = build_features(df)
print(df.select(["buy_ratio","sell_ratio","volume_delta","cvd","imbalance","volume_zscore"]).describe())
# CVD vs Close
sample = df.tail(5000).to_pandas()
fig, ax = plt.subplots(2,1, figsize=(12,6), sharex=True)
ax[0].plot(sample["timestamp"], sample["close"]); ax[0].set_title("Close")
ax[1].plot(sample["timestamp"], sample["cvd"]); ax[1].set_title("CVD")
plt.tight_layout(); plt.show()
```
